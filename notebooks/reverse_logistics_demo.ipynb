# Logistics Backtester — Reverse Logistics Demo

This notebook demonstrates generating a synthetic reverse-logistics dataset, running paired multi-seed experiments across current rules, OR-Tools optimization, and PPO, and inspecting statistical uplift.

```python
from app.config import load_app_settings
from app.data.synthetic import generate_synthetic_dataset
from app.data.validation import validate_dataset
from app.experiments.runner import run_backtest_suite
from app.experiments.scenarios import load_scenarios

settings = load_app_settings()
dataset = generate_synthetic_dataset(seed=42, horizon_minutes=2880.0, orders_per_day=90)
print(validate_dataset(dataset).model_dump())
```

```python
scenarios_map = load_scenarios()
selected_scenarios = [scenarios_map['baseline_normal'], scenarios_map['peak_surge_breakdown'], scenarios_map['storm_tight_sla']]
out = run_backtest_suite(
    dataset=dataset,
    scenarios=selected_scenarios,
    policy_ids=['current_replay', 'fifo_edf', 'greedy_cost', 'ortools_optimizer'],
    seeds=[101, 202, 303, 404, 505],
    settings=settings,
    experiment_name='notebook_demo',
    baseline_policy_id='current_replay',
)
out['comparison_df'].head(15)
```
